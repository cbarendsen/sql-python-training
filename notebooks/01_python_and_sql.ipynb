## 3. Let SQL do the heavy lifting, then chart it in Python

```python
query = """
SELECT c.CategoryName,
       SUM(oi.Quantity * oi.UnitPrice) AS SalesValue
FROM dbo.OrderItems AS oi
JOIN dbo.Orders     AS o ON o.OrderID    = oi.OrderID
JOIN dbo.Products   AS p ON p.ProductID  = oi.ProductID
JOIN dbo.Categories AS c ON c.CategoryID = p.CategoryID
WHERE o.Status <> 'Cancelled'
GROUP BY c.CategoryName
ORDER BY SalesValue DESC;
"""
sales = pd.read_sql(query, engine)
sales
```

```python
ax = sales.plot.bar(x="CategoryName", y="SalesValue", legend=False, rot=0)
ax.set_xlabel("")
ax.set_ylabel("Sales value (R)")
ax.set_title("Sales by category, 2026");
```
